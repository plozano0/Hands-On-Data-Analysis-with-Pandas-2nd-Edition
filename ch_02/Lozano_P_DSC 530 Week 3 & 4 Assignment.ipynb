Exercise 5(d) Sample Variance

Now we will focus on **Sample Variance**

We will begin by:

1. Count the observations *(passing as `n`)*
2. Take the sum of the `salaries` and divide by the `n` number of observations *(creating the `mean`)*
3. Obtain the **squared differences** by subtracting *each* observation by the `mean` and square it *(creating the `squared_diffs` variable)*
4. Lastly, we can get the variance by *summing* the `squared_diffs` variable and dividing it by the observations *(`n`)* - 1

```python
# Your 1st approach
# Count observations
n = len(salaries)
# Divide sum of salaries by n
mean = sum(salaries) / n
# Subtract mean from each salary, square the result, and sum all squared differences
squared_diffs = [(x - mean) ** 2 for x in salaries]
variance = sum(squared_diffs) / (n -1)

variance
```

Why do we subtract by 1 you ask?

This is because we are applying...

**[TERMINOLOGY ALERT]**
Degrees of Freedom (specifically Bessel's correction)

This means that we are compensating for a `1` degree of freedom due to the *sampling* of the data.

> **Note**:
>
> On a more complicated scale, we may have to increase our *'Degree of Freedom'* to ensure we are respecting the number of independent variables in the model. 
>
> This is done to ensure the estimate is unbiased.

```python
# Using statistics module
variance = statistics.variance(salaries)
variance
```

We can tell that **statistics** package is handling `variance()` as a sample variance since our results are the same.

> **Fun Fact**: 
> 
> **statistics** package has a module `pvariance()` that handles the situation I outlined above. 

If we are using the total population (In this case, we could infer this is true since we are dealing with random numbers),

we could negate the *-1* rule and simply divide by `n`.

"How significant of a difference is it", you say?

Let's find out!

```python
# Inserting 2nd approach for validation purposes
# Count observations
n = len(salaries)
# Divide sum of salaries by n
mean = sum(salaries) / n
# Subtract mean from each salary, square the result, and sum all squared differences
squared_diffs = [(x - mean) ** 2 for x in salaries]
# Divide sum of salaries by n - 1 to reflect statistics.variance behavior
# You subtract by 1 to get sample variance
# On more complicated contexts, you may need to use n - k - 1, where k is the number of independent variables.
variance_p = sum(squared_diffs) / n

variance_p
```

```python
difference_of_variance = variance - variance_p
difference_of_variance
```

Pretty significant if you ask me!

Exercise 5(e) Sample Standard Deviation

To calculate sample standard deviation:
1. We calculate the **Variance**
2. We take the square root of the variance

```python
# Your 1st approach
# Sample Standard Deviation calculation
n = len(salaries)
# Divide sum of salaries by n
mean = sum(salaries) / n
# Subtract mean from each salary, square the result, and sum all squared differences
squared_diffs = [(x - mean) ** 2 for x in salaries]
# Divide sum of squared differences by n - 1 to get sample variance
variance = sum(squared_diffs) / (n - 1)
# Take square root of variance to get standard deviation
std_dev = variance ** 0.5

std_dev
```

Using **statistics** package

```python
# Using statistics module
std_dev = statistics.stdev(salaries)
    
std_dev
```

Exercise 7(b) Standardizing

I will print both methods (**manual** and **scipy**) in 1 code block just for brevity purposes.

```python
# Refer to the Scaling data section of your reading in Chapter 1 of Hands-on Data Analysis with Pandas
# You can calculate this value by following the formula in the textbook or see the example in the solution
# You can also test out the z-score method in Scipy Stats (2D array is needed)

# Solution to display z-score standardized salaries (Standardization)
mean_salary = statistics.mean(salaries)
std_dev_salary = statistics.stdev(salaries)

# We subtract the mean from each salary and divide by the standard deviation to get the standardized value
standardized_salaries_manual = [(x - mean_salary) / std_dev_salary for x in salaries]

# Rounding to 8 decimal places to mimic scipy.stats.zscore behavior
standardized_salaries_manual = [round(x, 8) for x in standardized_salaries_manual]

# Testing with scipy stats zscore
from scipy import stats
salaries_reshaped = np.array(salaries).reshape(-1, 1)

# Really important to set ddof=1 to match sample standard deviation!
standardized_salaries = stats.zscore(salaries_reshaped, ddof=1)

# Printing both standardized salaries for comparison
print("Standardized Salaries (Manual Calculation):")
print(standardized_salaries_manual[:10])
print("\nStandardized Salaries (SciPy Stats zscore):")
print(standardized_salaries[:10])
```

**Note**: The standardized value is just a fancy way to say you are using **standard deviation** as a method to transform your data. This is indicated by the formula that is used to create it.

**Note**: The result from **scipy.stats.zscore()** with `ddof=0` will be calculating the population standard deviation as opposed to the sample.

Setting `ddof=1` makes it use sample standard deviation.

Exercise 8(a) The covariance between the standardized and normalized data

```python
# Covariance can be calculated with numpy
# When the (normalized) scaled_salaries variable was created, it was created as a 2D array.
# We will have to 'flatten' it to match the shape of the standardized_salary before calculation.
scaled_salaries = np.array(scaled_salaries).flatten()

# Now calculate the covariance
cov_calculated = np.cov(scaled_salaries, standardized_salary)
print(cov_calculated)
```

Note: How to interpret this output? The covariance function produce a 2X2 covariance matrix, the off-diagonal elements present the covariance of the 2 sets of values. Therefore 0.2685 is the output for intrepretation. Since the covariance estimate is positive, it indicates a positive relationship (moving in the same direction), and we don't know how "strong" this relationship and therefore correlation will help us understand this better.

```python
# Example 2 for covariance calculation
# Covariance can be calculated with statistics module
statistics.covariance(scaled_salaries, standardized_salary)
```

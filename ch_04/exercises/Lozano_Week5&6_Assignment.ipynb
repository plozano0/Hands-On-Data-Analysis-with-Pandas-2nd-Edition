## Exercise 1.
With the `earthquakes.csv` file, select all the earthquakes in Japan with a magnitude of 4.9 or greater using the `mb` magnitude type.

Importing necessary packages

```python
import pandas as pd
import numpy as np
```

Reading necessary csv's to complete Assignment

```python
# Reading both earthquakes and faang csv files
quakes = pd.read_csv('../../ch_04/exercises/earthquakes.csv')
faang = pd.read_csv('../../ch_04/exercises/faang.csv', index_col='date', parse_dates=True)
quakes.head()
```

```python
# We will filter the quakes dataframe for earthquakes in Japan with a magnitude of 4.9 or greater using the 'mb' magnitude type.
japan_quakes = quakes[(quakes['place'].str.contains('Japan')) &
                      (quakes['mag'] >= 4.9) &
                      (quakes['magType'] == 'mb')
                      ]
japan_quakes
```

# Exercise 2

Create bins for each full number of magnitude (*for example, the first bin is (0,1), the second is (1, 2), and so on*) with the `ml` magnitude type and count how many are in each bin.

```python
# Create bins of size 1 for the magnitude
mag_bins = np.arange(0,
                     quakes.loc[quakes['magType'] == 'ml']\
                         .mag.max() + 1
)

# Create a cross-tabulation of magnitude bins and magnitude types
binned_magnitudes = pd.crosstab(
    pd.cut(
        quakes.mag,
        bins=mag_bins
    ),
    quakes.magType
)

# Display only the 'ml' column
print(binned_magnitudes['ml'])
```

open and close prices with the **mean**,

the high price with the **max**,

the low price with the **min**,

# Exercise 7

Calculate the Z-scores of Amazon's data (ticker: AMZN) in Q4 2018 using apply().

```python
# Indexing the dataset
faang[
    (faang.index.quarter == 4) &
    (faang.index.year == 2018) &
    (faang['ticker'] == 'AMZN')
].drop(columns = 'ticker').apply(
    lambda x: x.sub(x.mean()).div(x.std())
).head()
```

# Exercise 10

The European Centre for Disease Prevention and Control (ECDC) provides an open dataset on COVID-19 cases called 

daily number of new reported cases of COVID-19 by country worldwide. 

This dataset is updated daily, but we will use a snapshot that contains data through September 18, 2020. 

Complete the following tasks to practice the skills you've learned up to this point in the book:

## Part 1

1. Read in the data in the `exercises/covid19_cases.csv` file
2. Create a `date` column by parsing the `dateRep` column into a datetime
3. Set the `date` column as the index
4. Use the `replace()` method to update all occurrences of `United_States_of_America` and `United Kingdom` to `USA` and `UK`, respectively
5. Sort the index

```python
# Assigning a date column by parsing the dateRep column into a datetime
# Setting the date column as the index
# Replacing long country names with their abbreviations
# Sorting the index
covid = pd.read_csv('../../ch_04/exercises/covid19_cases.csv')\
    .assign(date=lambda x: pd.to_datetime(x.dateRep, format='%d/%m/%Y'))\
    .set_index('date')\
    .replace('United_States_of_America', 'USA')\
    .replace('United_Kingdom', 'UK')\
    .sort_index()
```

## Part 2

For the 5 countries with the most cases (cumulative), find the day with the largest number of cases.

```python
# Creating a list of the top five countries with the most cumulative cases
top_five_countries = covid\
    .groupby('countriesAndTerritories').cases.sum()\
    .nlargest(5).index

# Indexing the covid dataframe for only the top five countries
# Finding the day with the largest number of cases for each of those countries
covid[covid.countriesAndTerritories.isin(top_five_countries)]\
    .groupby('countriesAndTerritories').cases.idxmax()
```

# Part 3

Find the 7-day average change in COVID-19 cases for the last week in the data for the countries found in Part 2.

```python
# Start with the 7-day rolling average of daily changes in cases for each country
rolling_avg = covid\
    .groupby(['countriesAndTerritories', pd.Grouper(freq='1D')]).cases.sum()\
    .unstack(0).diff().rolling(7).mean()

# Get the last date in the dataset
last_date = rolling_avg.index.max()

# Calculate the start date for the last 7 days
start_date = last_date - pd.Timedelta(days=6)

# Create a mask for the last 7 days
mask = (rolling_avg.index >= start_date) & (rolling_avg.index <= last_date)

# Apply the mask to get the results
result = rolling_avg.loc[mask, top_five_countries]
result
```

# Part 4

Find the first date that each country other than China had cases.

```python
# Using pivot to find the first date that each country other than China had cases
covid\
    .pivot(columns='countriesAndTerritories', values='cases')\
    .drop(columns='China')\
    .apply(lambda x: x[x > 0].index.min())\
    .sort_values()\
    .rename(lambda x: x.replace('_', ' '))
```

# Part 5

Rank the countries by total cases using percentiles.

```python
# Within the transform method, we will use rank with method='max' and pct=True to get the percentile rank of each country by total cases.
covid\
    .pivot_table(columns='countriesAndTerritories', values='cases', aggfunc='sum')\
    .T\
    .transform('rank', method='max', pct=True)\
    .sort_values('cases', ascending=False)\
    .rename(lambda x: x.replace('_', ' '))
```
